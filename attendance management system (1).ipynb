# ATTENDANCE MANAGEMENT SYSTEM USING PYTHON

```python
import pymysql
from datetime import date
```

```python
db = pymysql.connect(host="localhost", user="root" , database="attendance" , password = "1234")

cur = db.cursor()

def add_student():
    student_name = input("Enter student name: ")
    roll_no = int(input("Enter roll number: "))
    class_name = input("Enter class: ")

    query = "INSERT INTO students(student_name, roll_no, class) VALUES (%s, %s, %s)"
    values = (student_name, roll_no, class_name)

    cur.execute(query, values)
    db.commit()
    print("Student added successfully!")

def mark_attendance():
    student_id = int(input("Enter student ID: "))
    status = input("Enter status (Present/Absent/Leave): ")

    query = "INSERT INTO attendance(student_id, attendance_date, status) VALUES (%s, %s, %s)"
    values = (student_id, date.today(), status)

    cur.execute(query, values)
    db.commit()
    print("Attendance marked!")

def view_students():
    cur.execute("SELECT * FROM students")
    for row in cur.fetchall():
        print(row)

def view_attendance():
    student_id = int(input("Enter student ID: "))
    cur.execute("SELECT attendance_date, status FROM attendance WHERE student_id=%s", (student_id,))
    for row in cur.fetchall():
        print(row)

while True:
    print("\n--- Attendance Management System ---")
    print("1. Add Student")
    print("2. Mark Attendance")
    print("3. View Students")
    print("4. View Attendance of a Student")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        add_student()
    elif choice == 2:
        mark_attendance()
    elif choice == 3:
        view_students()
    elif choice == 4:
        view_attendance()
    elif choice == 5:
        print("Exiting...")
        break
    else:
        print("Invalid choice! Try again.")
```
